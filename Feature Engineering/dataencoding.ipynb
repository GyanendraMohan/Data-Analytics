# Data Encoding

        1. Nominal/OHE Encoding
        2. Label and ordinal encoding
        3. Target guided ordinal encodng

            Nominal/OHE Encoding:
                technique used to represent categorical data to numerical data, which is more suitable for machine learning algorithm. Each cataegory represented as binary vector where each bit corresponds to a unique category.

                1. Red[1,0,0]
                2. Green[0,1,0]
                3. Blue[0,0,1]

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'blue', 'red']
})  

df.head()
```

```python
#create an insatnce of OneHotEncoder
encoder = OneHotEncoder()
```

```python
# fit and transform the data
encoded_data = encoder.fit_transform(df[['color']]).toarray()
print(encoded_data)
```

```python
encoder_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(['color']))
encoder_df
```

```python
#for new data
encoder.transform([['blue']]).toarray()
```
