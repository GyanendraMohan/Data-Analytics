# Label Encoding
    Label encoding involves assigning a unique numericals label to each category in the variable. Assignment done in alphabetical order or based on the frequency of the category.

    Target labes with value between 0 and n_class_1

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'blue', 'red']
})  

df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df['color'])
```

```python
lbl_encoder.transform([['green']])
```

# Ordinal encoding
        Used to encode categorical data that have an intrinic order or ranking, in this each category assigned numerical value based on its position in the order.

        Example:
            1. High School == 1
            2. College == 2
            3. Graduate == 3
            4. Post-Graduate == 4
