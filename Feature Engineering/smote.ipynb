# SMOTE - (synthetic minority oversampling technique)

        - This is used in machine learning to address imballanced dataset where the minorty class has significantly fewer instance then the majority class. SMOTE inovolves generating synthetic instances of minority class by interpolating between the existing instances.

```python
from sklearn.datasets import make_classification # this help to create a dataset 
```

```python
make_classification(
    n_samples = 1000,
    n_features = 2,
    *,
    n_informative = 2,
    n_redundant = 2,
    n_repeated = 0,
    n_classes = 2,
    n_clusters_per_class = 2,
    weights = None,
    flip_y = 0.01,
    class_sep = 1.0,
    hypercube = True,
    shift = 0.0,
    scale = 1.0,
    shuffle = True,
    random_state = 42
)
```
