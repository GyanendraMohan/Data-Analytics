# Upsampling and Downsampling

# 1 Upsampling -  replace is TRUE 
# 2 Downsampling - replace is FALSE because we want to reduce the data, downsampling is not good as we are loosing the data.

```python
import numpy as np
import pandas as pd 

np.random.seed(123)

n_samples = 1000
class_0_ration = 0.9
n_class_0 = int(n_samples * class_0_ration)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
#create my df with imbalanced dataset

class_0 = pd.DataFrame({
    'feature1': np.random.normal(loc=0.0, scale=1.0, size=n_class_0),
    'feature2': np.random.normal(loc=5.0, scale=2.0, size=n_class_0),
    'target': np.zeros(n_class_0)
})  

class_1 = pd.DataFrame({
    'feature1': np.random.normal(loc=2.0, scale=1.0 , size=n_class_1),
    'feature2': np.random.normal(loc=7.0, scale=2.0, size=n_class_1),
    'target': np.ones(n_class_1)
})  

print(class_0, class_1)
```

```python
df = pd.concat([class_0, class_1], axis=0).reset_index(drop=True)
```

```python
df.head()
```

```python
df_minority = df[df.target == 1]
df_majority = df[df.target == 0]
```

```python
from sklearn.utils import resample #it extrapolates the minority class, ie it creates new samples from the existing samples by random sampling with replacement
```

```python
df_minority_upsampled = resample(df_minority, replace=True, n_samples=len(df_majority), random_state=42)
```

```python
df_minority_upsampled.shape
```

```python
df_upsampled=pd.concat([df_majority, df_minority_upsampled])
df_upsampled
```
