# Lab: Handling Missing Values

## Introduction

Missing values in a dataset occur when some of the information is not stored for a variable. There are three mechanisms for missing data:

### 1. Missing Completely at Random (MCAR)
- Probability of a value being missing is unrelated to both the observed and the missing data.
- Missing values are randomly distributed over the dataset.
- There is no systematic reason for this.

### 2. Missing at Random (MAR)
- Probability of the value depends only on the observed data, but not on the missing data itself.

### 3. Missing Data Not at Random (MNAR)
- Probability of missing data depends on the missing data itself.
- If the data is MNAR, the missingness is not random and is dependent on unobserved or unmeasured factors that are associated with the missing value.


```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the Titanic dataset
df = sns.load_dataset('titanic')
df.head()
```

## Step 1: Check Missing Values


```python
# Check missing values
df.isnull().sum()
```

```python
# Check the shape of the dataset
df.shape
```

## Step 2: Deletion Methods (Not Recommended)

### Row-wise Deletion
**Problem:** We may lose a lot of data. This is a blunder and should be avoided.


```python
# Delete rows with missing values
df.dropna().shape
```

### Column-wise Deletion


```python
# Column-wise deletion
df.dropna(axis=1).shape
```

## Step 3: Imputation Techniques

### 1. Mean Value Imputation
- Works well when we have normally distributed data.


```python
# Visualize the distribution of age
sns.histplot(df['age'], kde=True)
plt.title('Distribution of Age')
plt.show()
```

```python
# Mean imputation
df['Age_mean'] = df['age'].fillna(df['age'].mean())
```

```python
# Compare original and imputed values
df[['age', 'Age_mean']]
```

### 2. Median Value Imputation
- Useful when we have outliers in the dataset.


```python
# Median imputation
df['Age_median'] = df['age'].fillna(df['age'].median())
```

```python
# Compare all three columns
df[['Age_median', 'Age_mean', 'age']]
```

### 3. Mode Imputation Technique
- Best suitable for categorical values.


```python
# Check rows with missing embarked values
df[df['embarked'].isnull()]
```

```python
# Check unique values in embarked column
df['embarked'].unique()
```

```python
# Calculate mode value (all NaN values get skipped)
mode_value = df[df['embarked'].notna()]['embarked'].mode()[0]
print(f"Mode value for embarked: {mode_value}")
```

```python
# Mode imputation
df['Embarked_mode'] = df['embarked'].fillna(mode_value)
```

```python
# Compare original and imputed values
df[['Embarked_mode', 'embarked']]
```

```python
# Verify no missing values in imputed column
print(f"Missing values in Embarked_mode: {df['Embarked_mode'].isnull().sum()}")
print(f"Missing values in original embarked: {df['embarked'].isnull().sum()}")
```

## Summary

This lab demonstrated various techniques for handling missing values:

1. **Deletion Methods** (Not Recommended):
   - Row-wise deletion: Removes entire rows with missing values
   - Column-wise deletion: Removes entire columns with missing values
   - Both methods result in significant data loss

2. **Imputation Techniques**:
   - **Mean Imputation**: Suitable for normally distributed numerical data
   - **Median Imputation**: Better for data with outliers
   - **Mode Imputation**: Best for categorical data

The choice of imputation method depends on the nature of the data and the analysis requirements.
