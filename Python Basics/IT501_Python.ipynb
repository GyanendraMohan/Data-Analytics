# ARRAY CREATION 

The array module cannot store strings.
It only stores numeric types and characters, depending on a type code.

from array import array

import array

Syntax

array_name = array.array(‘type_code’, [elements])


Type Codes (All Supported Types)
| Type Code | Python Type       | Size         | Description          |
| --------- | ----------------- | ------------ | -------------------- |
| `'b'`     | signed char       | 1 byte       | Integers −128 to 127 |
| `'B'`     | unsigned char     | 1 byte       | 0–255                |
| `'u'`     | Unicode character | 2 or 4 bytes | Characters           |
| `'h'`     | signed short      | 2 bytes      |                      |
| `'H'`     | unsigned short    | 2 bytes      |                      |
| `'i'`     | signed int        | 2 or 4 bytes |                      |
| `'I'`     | unsigned int      | 2 or 4 bytes |                      |
| `'l'`     | signed long       | 4 bytes      |                      |
| `'L'`     | unsigned long     | 4 bytes      |                      |
| `'f'`     | float             | 4 bytes      |                      |
| `'d'`     | double            | 8 bytes      |                      |



```python
# Integer Array ('i')
from array import array

arr = array('i', [1, 2, 3, 4])
print(arr)        


# Float Array ('f')
arr = array('f', [1.1, 2.2, 3.3])
print(arr)


# Double Array ('d')
arr = array('d', [1.234, 5.678])
print(arr)


# Character Array ('u')
arr = array('u', ['a', 'b', 'c'])
print(arr)

arr = array('u', '1234h')
print(arr)    




```

```python
# ARRAY OPERATIONS
# Accessing Elements
from array import array
arr = array('i', [10, 20, 30])
print(arr[0])   


# Modifying Elements
arr = array('i', [10, 20, 30])
arr[1] = 99
print(arr)  


# Append Items
arr = array('i', [1, 2, 3])
arr.append(4)
print(arr)


# Insert Items
arr = array('i', [1, 2, 3])
arr.insert(1, 100)
print(arr)


# Remove Items
arr = array('i', [1, 2, 3, 2])
arr.remove(2)  
print(arr)  

# arr = array('i', [10, 20, 30])
arr.pop(1)
print(arr)
arr.pop()
print(arr)

# Looping
arr = array('i', [1, 2, 3])
for item in arr:
    print(item)


# Array Length
print(len(arr))


#Converting List to Array
lst = [1, 2, 3]
arr = array('i', lst)

# Converting Array to List
arr = array('i', [1, 2, 3])
lst = arr.tolist()
print(lst)


# Slice an Array
arr = array('i', [10, 20, 30, 40])
print(arr[1:3])  


# Extend with another array
a = array('i', [1, 2])
b = array('i', [3, 4])
a.extend(b)
print(a)



from array import array
arr = array('u', ["apple", "banana"])  

arr = array('i', [1,2])
print(arr)







```

```python
# Concat operation in data frame 

df1 = pd.DataFrame({
    "A": [1, 2],
    "B": [3, 4]
})

df2 = pd.DataFrame({
    "A": [5, 6],
    "B": [7, 8]
})

result = pd.concat([df1, df2], axis=0)
print(result)

result = pd.concat([df1, df2], axis=0, ignore_index=True)
print(result)

# Using Different Join Types
# Outer join (default),  union of columns, missing values filled with NaN
# Inner join, intersection of columns only

df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"B": [5, 6], "C": [7, 8]})

outer = pd.concat([df1, df2], axis=0, join='outer')
inner = pd.concat([df1, df2], axis=0, join='inner')

print("Outer Join:\n", outer)
print("\nInner Join:\n", inner)

```

# Merge operation
pd.merge(left, right, 
         how='inner', 
         on=None, 
         left_on=None, 
         right_on=None,
         left_index=False, 
         right_index=False)

| Parameter                   | Description                                                   |
| --------------------------- | ------------------------------------------------------------- |
| `left`, `right`             | DataFrames to merge                                           |
| `how`                       | Type of merge: `'inner'`, `'outer'`, `'left'`, `'right'`      |
| `on`                        | Column(s) common to both DataFrames to merge on               |
| `left_on`, `right_on`       | Column(s) to merge from left and right if column names differ |
| `left_index`, `right_index` | Merge on index instead of columns                             |


| `how`     | Description                                                              |
| --------- | ------------------------------------------------------------------------ |
| `'inner'` | Only include rows with matching keys in both DataFrames (default)        |
| `'outer'` | Include all rows from both DataFrames, fill missing values with `NaN`    |
| `'left'`  | Include all rows from left DataFrame, fill missing from right with `NaN` |
| `'right'` | Include all rows from right DataFrame, fill missing from left with `NaN` |


```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"]
})

df2 = pd.DataFrame({
    "ID": [2, 3, 4],
    "Age": [25, 30, 35]
})

# Merge on "ID" column
result = pd.merge(df1, df2, on="ID", how="inner")
print(result)
```
