# COVID-19 Patients Data Analysis

This notebook analyzes the `covid19_dataset.csv` file.

**Tasks:**
1. Load `covid19_dataset.csv`.
2. Drop rows where `age` or `id` is missing.
3. Replace missing `symptom_onset` with the most frequent value.
4. Sort patients by `date_onset_symptoms` (ascending) and then by `age` (descending).

```python
import pandas as pd

# 1. Load covid19_dataset.csv
df = pd.read_csv('COVID19_dataset.csv')

# Display the first few rows
df.head()
```

```python
# 2. Drop rows where age or id is missing
df_cleaned = df.dropna(subset=['age', 'id']).copy()

# Verify the shape after dropping
print(f"Original shape: {df.shape}")
print(f"Cleaned shape: {df_cleaned.shape}")
```

```python
# 3. Replace missing symptom_onset with the most frequent value
most_frequent_symptom = df_cleaned['symptom_onset'].mode()[0]
df_cleaned['symptom_onset'] = df_cleaned['symptom_onset'].fillna(most_frequent_symptom)

print(f"Most frequent symptom: {most_frequent_symptom}")
# check if any nulls remain
df_cleaned.isnull().sum()
```

```python
# 4. Sort patients by date_onset_symptoms (ascending) and then by age (descending)
# First ensure date_onset_symptoms is datetime objects for correct sorting
df_cleaned['date_onset_symptoms'] = pd.to_datetime(df_cleaned['date_onset_symptoms'])

df_sorted = df_cleaned.sort_values(by=['date_onset_symptoms', 'age'], ascending=[True, False])

# Display the result
df_sorted
```
