# Iris Dataset Analysis

This notebook analyzes the Iris dataset using Seaborn visualizations.
We will perform the following steps:
1. Load the dataset.
2. Generate a Pair Plot of all features.
3. Create a Correlation Heatmap of numerical features.
4. Create a Boxplot of petal length per species.
5. Derive insights from the visualizations.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set theme for better aesthetics
sns.set_theme(style="whitegrid")
%matplotlib inline
```

## 1. Load Iris Dataset

```python
# Load the iris dataset using seaborn
df = sns.load_dataset('iris')

# Display first few rows to verify
df.head()
```

## 2. Visualizations

### Pair Plot of All Features

```python
# Pair plot colored by species
sns.pairplot(df, hue='species', height=2.5)
plt.suptitle('Pair Plot of Iris Dataset', y=1.02)
plt.show()
```

### Correlation Heatmap of Numerical Features

```python
# Select only numerical columns for correlation calculation
numeric_df = df.select_dtypes(include=[np.number])

# Calculate correlation matrix
corr = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

### Boxplot of Petal Length per Species

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='species', y='petal_length', data=df)
plt.title('Distribution of Petal Length across Species')
plt.show()
```

## 3. Insights

Based on the visualizations above, here are two key insights:

1.  **Strong Positive Correlation:** The heatmap and pair plot reveal a very strong positive correlation between `petal_length` and `petal_width` (correlation coefficient close to 1). This indicates that as the petal length increases, the petal width also tends to increase proportionally.

2.  **Species Separation:** The `setosa` species is clearly distinguishable from `versicolor` and `virginica`. In the pair plot and boxplot, `setosa` forms a distinct cluster with significantly smaller petal lengths and widths. While `versicolor` and `virginica` have some overlap, `virginica` generally has larger petal dimensions than `versicolor`.
