# PageRank Algorithm Assignment

**Instructions to Run:**

1. Execute the notebook sequentially, starting from the first code cell.
2. When prompted, enter the total number of nodes in your graph (recommended: try at least 3).
3. For each node, enter a space-separated row describing its outgoing connections:
    - Example (Chain graph 3 nodes): 
      - Row 1: `0 1 0`
      - Row 2: `0 0 1`
      - Row 3: `0 0 0`
4. Choose the execution mode:
    - Type `1` for a fixed number of iterations
    - Type `2` to run until convergence (recommended for this assignment)
5. If you select convergence mode, enter the tolerance (for example, `1e-6`).
6. The output will display the PageRank vector and number of iterations required for convergence.

---

**Tip:** Use diverse adjacency matrices for testing (e.g., a linear chain, loops, star, or a random web) to observe how many iterations are needed!


```python
import numpy as np  # pyright: ignore[reportMissingImports]

def pagerank_iteration(adj_matrix, pr_vector, d=0.85):
    N = len(adj_matrix)
    stochastic_matrix = adj_matrix / adj_matrix.sum(axis=0, keepdims=True)
    stochastic_matrix = np.nan_to_num(stochastic_matrix)
    pr_new = d * stochastic_matrix.dot(pr_vector) + (1 - d) / N
    return pr_new


def pagerank_fixed_iterations(adj_matrix, pr_init, iterations):
    pr = pr_init.copy()
    for _ in range(iterations):
        pr = pagerank_iteration(adj_matrix, pr)
    return pr

def pagerank_until_convergence(adj_matrix, pr_init, tolerance=1e-6, max_iter=1000):
    pr = pr_init.copy()
    for i in range(max_iter):
        new_pr = pagerank_iteration(adj_matrix, pr)
        if np.linalg.norm(new_pr - pr, ord=1) < tolerance:
            print(f"Converged in {i+1} iterations")
            return new_pr
        pr = new_pr
    print("Did not fully converge within max iterations")
    return pr

def initialize_vector(N):
    print("Using uniform initialization")
    return np.ones(N) / N

```

```python
print("=== PAGE RANK COMPUTATION ===\n")

N = int(input("Enter number of nodes N: "))

print("\nEnter adjacency matrix row by row.")
print("Use 1 for a link, 0 for no link.\n")

adj_matrix = []
for i in range(N):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    adj_matrix.append(row)

adj_matrix = np.array(adj_matrix, dtype=float)

print("\nInitialization Method:")
print("1. Uniform (1/N)")

pr = initialize_vector(N)

print("\nExecution Mode:")
print("1. Run for fixed number of iterations")
print("2. Run until convergence")

mode = int(input("\nChoose execution mode (1-2): "))

if mode == 1:
    iterations = int(input("Enter number of iterations: "))
    result = pagerank_fixed_iterations(adj_matrix, pr, iterations)
    print("\nResult after fixed iterations:\n", result)

elif mode == 2:
    tol = float(input("Enter convergence tolerance (e.g., 1e-6): "))
    result = pagerank_until_convergence(adj_matrix, pr_init=pr, tolerance=tol)
    print("\nConverged PageRank Vector:\n", result)
else:
    print("Invalid option")
```
